### Sales Data 

### Problem Statement
The retail company is preparing its monthly sales performance report, but the raw sales data contains missing values, duplicate records, inconsistent formats, incorrect data types, invalid numerical values, and potentially incorrect Sales and Profit values. If these issues are not resolved before analysis, management may make decisions using inaccurate revenue, profitability, product, regional, and salesperson information. Your task is to act as a Data Analyst and convert the raw sales file into a reliable, consistent, and analysis-ready dataset that can be safely used for reporting and future EDA.



### Objective of the Data
1. Data Loading & Initial Inspection
2. Missing Values & Blank Data
3. Duplicate Records & Order IDs
4. Data Type & Format Correction
5. Text Standardization
6. Invalid Value Detection
7. Business Rule Validation
8. Create the Cleaned Dataset
9.  Final Data Quality Validation
10. Export & Document the Result  

```python
Data Loading & Initial Inspection — Load the raw CSV into Pandas and inspect its shape, columns, sample records, data types, and overall structure.
2. Missing Values & Blank Data — Identify missing values, empty strings, and whitespace-only values, then determine an appropriate treatment for each affected field.
3. Duplicate Records & Order IDs — Detect exact duplicate rows and duplicate Order_ID values, investigate the affected records, and decide which records should remain.
4. Data Type & Format Correction — Correct incorrect or mixed data types and standardize Order_Date, Quantity, Unit_Price, Discount, Sales, and Profit into appropriate formats.
5. Text Standardization — Clean and standardize Region, Product, Category, and Salesperson values by handling spaces, capitalization differences, and inconsistent labels.
6. Invalid Value Detection — Identify negative, impossible, or out-of-range values in Quantity, Unit_Price, Discount, Sales, and Profit and decide how each issue should be resolved.
7. Business Rule Validation — Check whether Sales is logically consistent with Quantity, Unit_Price, and Discount, and investigate suspicious Profit values using an appropriate business rule.
8. Create the Cleaned Dataset — Apply all justified cleaning decisions to a copy of the raw dataframe while preserving the original raw CSV unchanged.
9. Final Data Quality Validation — Recheck missing values, duplicates, data types, invalid values, unique Order_IDs, and category consistency to confirm that the cleaned dataset is analysis-ready.
10. Export & Document the Result — Save the final dataset as Sales Data Cleaned.csv and prepare a concise before-versus-after data-quality summary documenting the major corrections.
```
